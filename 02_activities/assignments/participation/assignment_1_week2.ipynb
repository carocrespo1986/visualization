```python
np.random.seed (613)
x1 = np.arange(50)
y1 = np.random.randint(0,75,50)
x2 = np.array (["Luffy", "Zorro", "Nami", "Usopp", "Sanji"])
y2 = np.array([100,180,240,99,220])
```

```python
fig, (ax1, ax2) = plt.subplots(ncols = 2,
                               nrows = 1,
                               figsize = (7,3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
```

```python
font1 = {'family':'cursive', 'weight' : 'bold'}
font2 = {'color':'green', 'weight': 'bold', 'family':'arial'}
fig, (ax1, ax2) = plt.subplots(ncols = 1,
                               nrows = 2,
                               figsize = (3,7))
ax1.scatter(x1,y1, color = 'orange')
ax1.set_ylabel('Axis Y', fontdict=font2)
ax1.set_title('Scatter Plot', fontdict=font1)

ax2.barh(x2,y2)
ax2.set_ylabel('Characters', fontdict=font2)
ax2.set_title('Bar Chart', fontdict=font1)
```

```python
fig, someaxes = plt.subplot_mosaic ([['ax1','ax3'],
                                     ['ax2','ax3']],
                                     figsize = (7,3))
someaxes['ax1'].scatter(x1, y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)
```

```python
fig, someaxes = plt.subplot_mosaic ([['ax1','ax3'],
                                     ['ax2','ax3']],
                                     figsize = (7,3),
                                     layout = "constrained") #makes everything fit
someaxes['ax1'].scatter(x1, y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)

someaxes['ax1'].set_xlabel('A big Label', fontsize =18)
someaxes['ax2'].set_xlabel('Another Label', fontsize=18)
someaxes['ax3'].set_xlabel('Label 2: 2 fast 2 furious', fontsize = 18)
```

```python
x = np.array(["Luffy", "Zoro","Nami","Usopp","Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

fig, ax = plt.subplots(figsize=(7, 3))

ax.bar(x, y1, color = "indigo")
ax.plot(x, y2, color = "red")
```

```python
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize =(7,3))

ax.plot(x, y2, color = "red", linestyle = "--")

ax.errorbar(x, y2,
            yerr = y2_sd,
            fmt= "none", 
            color = "orange", #if no fmt, ecolor is needed
            capsize = 6,
            linewidth = 4,
            capthick = 6,
            errorevery = 2
            ) #errorbar is the way of adding those lines
```

```python
#for including images in the graphs
from PIL import Image
import requests
from io import BytesIO

response = requests.get('https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png')
image_file = BytesIO (response.content)
image = Image.open(image_file)
```
